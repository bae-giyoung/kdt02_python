```python
file_path = 'file_datas/weather_data/sitka_weather_2021_simple.csv'
with open(file_path, 'r') as f:
    reader = csv.DictReader(f)
    header = next(reader)
    print('header', header)
    result = []
    for line in reader:
        print(line.keys())
        result["STATION"] = result.get('STATION', []) + [line['STATION']]
        result["NAME"] = result.get('NAME', []) + [line['NAME']]
        result["DATE"] = result.get('DATE', []) + [line['DATE']]
        result["TAVG"] = result.get('TAVG', []) + [line['TAVG']]
        result["TMAX"] = result.get('TMAX', []) + [line['TMAX']]
        result["TMAX"] = result.get('TMAX', []) + [line['TMAX']]
        result["STATION"].append((line["STATION"], line["NAME"], line["DATE"], line["TAVG"], line["TMAX"], line["TMIN"]))

result
```
