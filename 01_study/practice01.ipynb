```python
import numpy as np
import matplotlib.pyplot as plt

def f1(x):
    return 2*x + 1

def f2(x):
    return x**2 + x*2 + 1

#x = np.arange(-5,6) # 각진 느낌
x = np.linspace(-6, 5, 100) # 100개로 쪼갬
y1 = f1(x)
y2 = f2(x)
plt.plot(x, y1, c="red", ls="--", lw='1', label='f1(x)')
plt.plot(x, y2, c="blue", ls="-", lw='1', label='f2(x)')
plt.title('Two Graphs')
plt.xlabel('X')
plt.ylabel('Y')
plt.grid()
plt.legend()
```

```python
from random import choice

class RandomWalk:
    """랭덤 워크를 만드는 클래스"""
    def __init__(self, num_points=5000): #
        """속성을 초기화합니다"""
        self.num_points = num_points

        # 이동은 (0,0)에서 시작
        self.x_values = [0]
        self.y_values = [0]

    def fill_walk(self):
        """랜덤 워크의 각 포인트를 계산합니다"""

        # 설정한 이동 수에 도달할 때까지 움직임을 반복
        while len(self.x_values) < self.num_points:

            #방향과 거리를 정함
            x_direction = choice([1,-1]) # random.choice()
            x_distance = choice([0,1,2,3,4])
            x_step = x_direction * x_distance

            y_direction = choice([1,-1])
            y_distance = choice([0,1,2,3,4])
            y_step = y_direction * y_distance

            #움직임이 없는 결정은 버림
            if x_step == 0 and y_step == 0:
                continue

            #새 위치를 계산
            x = self.x_values[-1] + x_step
            y = self.y_values[-1] + y_step

            self.x_values.append(x)
            self.y_values.append(y)
```

```python
import matplotlib.pyplot as plt

# 랜덤 워크를 생성
rw = RandomWalk()
rw.fill_walk()

# 랜덤 워크의 포인트를 그림
plt.style.use('seaborn-v0_8-pastel')
fig, ax = plt.subplots()
ax.scatter(rw.x_values, rw.y_values, s=10)
ax.set_aspect('equal')
plt.show()
```
