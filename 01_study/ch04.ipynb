CSV 파일

```python
import csv

with open('grade.csv', 'r') as f:
    file = csv.reader(f)
    students = {}
    subjects = {}
    headers = next(file) # 첫번째 라인 날아가고 나서
    for line in file:
        #print(line)
        name, java, pyth = line[0], int(line[1]), int(line[2])
        students[name] = sum([java, pyth])/2
        subjects['java'] = subjects.get('java', 0) + java
        subjects['python'] = subjects.get('python', 0) + pyth
students
```

```python
# 학생별 평균

# 점수별 평균
for k, v in subjects.items():
    print(k, v/len(subjects))
```

```python
import json

lst = [1,2,3]

content = json.dumps(lst) # '[1, 2, 3]' serialize
print(type(content), content)

request =json.loads(content) # [1, 2, 3] unserialize
print(type(request) ,request)
```

```python
data = {
    'Kim': [90,80],
    'Lee': [77,88],
    'Park': [60,50],
}

data2 = { # col로 
    'name': ['Kim','Lee','Park'],
    'java': [90,77,60],
    'python': [80,88,50],
}

response = json.dumps(data) # '{"Kim": [90, 80], "Lee": [77, 88], "Park": [60, 50]}'

request = json.loads(response) # {'Kim': [90, 80], 'Lee': [77, 88], 'Park': [60, 50]}
print(type(request), request)
```

```python
with open('data.json', 'w') as f:
    json.dump(data2, f)

with open('data.json', 'r') as f:
    data4 = json.load(f)

data
```
